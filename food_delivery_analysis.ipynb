```python
raw_df = pd.read_json("India-Food-Delivery-Time-Prediction.txt")

df["Time_Order_picked"] = raw_df["Time_Order_picked"].copy()
```

```python
df["Time_Order_picked"].head()
```

```python
df["Delivery_person_Age"] = pd.to_numeric(
    df["Delivery_person_Age"], errors="coerce"
)
df["Delivery_person_Age"] = df["Delivery_person_Age"].fillna(
    df["Delivery_person_Age"].median()
)

# Ratings
df["Delivery_person_Ratings"] = pd.to_numeric(
    df["Delivery_person_Ratings"], errors="coerce"
)
df.loc[df["Delivery_person_Ratings"] > 5, "Delivery_person_Ratings"] = pd.NA
df["Delivery_person_Ratings"] = df["Delivery_person_Ratings"].fillna(
    df["Delivery_person_Ratings"].median()
)

# Order Date
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"], format="%d-%m-%Y", errors="coerce"
)

# Time Ordered
df["Time_Orderd"] = pd.to_datetime(
    df["Time_Orderd"], format="%H:%M:%S", errors="coerce"
)

# Time Order Picked — keep original time format
df["Time_Order_picked"] = df["Time_Order_picked"].astype("string").str.strip()

# Multiple deliveries
df["multiple_deliveries"] = pd.to_numeric(
    df["multiple_deliveries"], errors="coerce"
)
df["multiple_deliveries"] = df["multiple_deliveries"].fillna(
    df["multiple_deliveries"].median()
)

# Time taken
df["Time_taken(min)"] = (
    df["Time_taken(min)"]
    .astype(str)
    .str.replace("(min) ", "", regex=False)
)
df["Time_taken(min)"] = pd.to_numeric(
    df["Time_taken(min)"], errors="coerce"
)
```
